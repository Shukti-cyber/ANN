# Text Cleaning
----------------------------------
**1. Remove punctuation**

**2. Remove stopwords**

**3. Stemming/ Lemmatization**

```python
import pandas as pd
df = pd.DataFrame({"in":["He is a good good boy.","She is a good girl.","boy & girl are bad."],"op":["P","P","N"]})
df
```

```python
df["in"][2]
```

```python
corpus=[]
import re
for i in range(len(df)):
    rp = re.sub('[^a-zA-Z]'," ",df["in"][i])
    rp = rp.lower()
    rp = rp.split()
    rp = [ps.stem(word) for word in rp if not word in stopwords.words('english')]
    rp = " ".join(rp)
    corpus.append(rp)

pd.DataFrame({"in":corpus})
```

# Vectorization

**Count Vectorizer (Bag of Words)** --> it counts the repetion no. of words in the given sentence

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()

pd.DataFrame(cv.fit_transform(corpus).toarray(),columns=cv.get_feature_names_out())
```

```python
pd.DataFrame(cv.fit_transform(corpus).toarray(),columns=cv.get_feature_names_out())
```

**TF-IDF Vectorizer**

- Term Frequency (tf) = No. of times term ‘t’ occurs in a document


- Inverse Document Frequency (idf) is a measure of how common or rare a term is across the entire corpus of documents. So the point to note is that it’s common to all the documents. If the word is common and appears in many documents, the idf value (normalized) will approach 0 or else approach 1 if it’s rare.

idf(word) = 1 + ln [n/df(t)]

where

n = Total number of documents available

t = term for which idf value has to be calculated

df(t) = Number of documents consists of given word

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf= TfidfVectorizer()
tf.fit_transform(corpus).toarray()
```

```python
tf.get_feature_names_out()
```
