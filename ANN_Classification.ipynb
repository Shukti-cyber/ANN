```python
x=pd.get_dummies(df.drop('Exited',axis=1),drop_first=True)
y=df['Exited']
```

```python
### Splitting the dataset into the Training sett and test set
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

Modelling

```python
from keras.models import Sequential
ann= Sequential()
```

```python
from keras.layers import Dense
ann.add(Dense(input_shape=(11,),units=6,kernel_initializer='uniform',activation='relu'))
```

```python
ann.add(Dense(units=6,kernel_initializer='uniform',activation='relu'))
```

```python
ann.add(Dense(units=1,kernel_initializer='uniform',activation='sigmoid'))
```

```python
ann.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
ann.fit(X_train,y_train,epochs=100,batch_size=32)
```

Prediction and evaluating the model


```python
y_pred=ann.predict(X_test)
y_pred=(y_pred>0.5)
```

```python
### Splitting the dataset into the Training sett and test set
from sklearn.metrics import confusion_matrix,accuracy_score

print("Test Accuracy :",accuracy_score(y_test,y_pred))
confusion_matrix(y_test,y_pred)
```

```python
def build_cross_classifier():
  classifier=Sequential()
  classifier.add(Dense(input_shape=(11,),units=6,activation='relu',kernel_initializer='uniform'))
  classifier.add(Dense(units=6,activation='relu',kernel_initializer='uniform'))
  classifier.add(Dense(units=1,activation='sigmoid',kernel_initializer='uniform'))
  classifier.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
  return classifier
```

```python
from scikeras.wrappers import KerasClassifier, KerasRegressor
from scikeras.wrappers import KerasClassifier
classifier = KerasClassifier(build_cross_classifier,batch_size=32,epochs=100)
```

```python
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
import numpy as np

# Standardize cross-validation to bypass scikeras compatibility bugs with scikit-learn 1.6+
kf = KFold(n_splits=5, shuffle=True, random_state=0)
accuracies = []

# Convert to numpy arrays for indexing
X_arr = x.values
y_arr = y.values

for fold, (train_idx, val_idx) in enumerate(kf.split(X_arr, y_arr)):
    X_tr, X_val = X_arr[train_idx], X_arr[val_idx]
    y_tr, y_val = y_arr[train_idx], y_arr[val_idx]

    # Neural networks need input feature scaling per fold
    scaler = StandardScaler()
    X_tr_scaled = scaler.fit_transform(X_tr)
    X_val_scaled = scaler.transform(X_val)

    # Rebuild and train a fresh model for each fold
    fold_classifier = build_cross_classifier()
    fold_classifier.fit(X_tr_scaled, y_tr, epochs=100, batch_size=32, verbose=0)

    # Evaluate fold accuracy
    loss, accuracy = fold_classifier.evaluate(X_val_scaled, y_val, verbose=0)
    accuracies.append(accuracy)
    print(f"Fold {fold + 1} Accuracy: {accuracy:.4f}")

print(f"\nMean CV Accuracy: {np.mean(accuracies):.4f}")
```

```python
estimator = KerasClassifier(build_cross_classifier())
param_grid= { 'batch_size':[10,32],'epoch':[50,100],"optimizer":['adam','rmsprop']}
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator,param_grid, scoring='accuracy',cv=5)
grid_result=grid.fit(X_train,y_train)
```
